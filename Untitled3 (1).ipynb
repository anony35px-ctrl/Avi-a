```python
# Importar las librerías necesarias
import pandas as pd

# Cargar el dataset
datos = pd.read_csv("housing.csv")
```

```python
# Mostrar las primeras filas para inspeccionar la estructura
datos.head()
```

```python
# Contar los valores de la columna categórica 'ocean_proximity'
datos["ocean_proximity"].value_counts()
```

```python
# Obtener información general del DataFrame (tipos de datos y valores nulos)
datos.info()
```

```python
# Estadísticas descriptivas de las columnas numéricas
datos.describe()
```

```python
# Histogramas de todas las variables numéricas
datos.hist()
```

```python
# Gráfico de dispersión de longitud vs latitud para visualizar la distribución geográfica
import seaborn as sb
sb.scatterplot(x="longitude", y="latitude", data=datos)
```

```python
# Filtrar las viviendas con ingresos medios superiores a 14 y graficar con color según el valor de la vivienda
sb.scatterplot(x="longitude", y="latitude", data=datos[(datos.median_income > 14)], hue="median_house_value", palette="coolwarm")
```

```python
# Verificar nuevamente la información del DataFrame
datos.info()
```

```python
# Eliminar filas con valores nulos (solo afecta a 'total_bedrooms')
datos_na = datos.dropna()
```

```python
# Confirmar que ya no hay valores nulos
datos_na.info()
```

```python
# Visualizar la columna 'ocean_proximity' en el DataFrame limpio
datos_na["ocean_proximity"]
```

```python
# Contar frecuencias de 'ocean_proximity' después de la limpieza
datos_na["ocean_proximity"].value_counts()
```

```python
# Crear variables dummy (one-hot encoding) para la columna categórica
dummies = pd.get_dummies(datos_na["ocean_proximity"], dtype=int)
```

```python
# Unir las variables dummy al DataFrame principal
datos_na = datos_na.join(dummies)
```

```python
# Ver las primeras filas con las nuevas columnas
datos_na.head()
```

```python
# Eliminar la columna original 'ocean_proximity' para evitar redundancia
datos_na = datos_na.drop(["ocean_proximity"], axis=1)
```

```python
# Confirmar la eliminación visualizando las primeras filas
datos_na.head()
```

```python
# Calcular la matriz de correlación de todas las variables numéricas
datos_na.corr()
```

```python
# Configurar el tamaño de la figura y dibujar un mapa de calor de las correlaciones
sb.set(rc={'figure.figsize':(15,8)})
sb.heatmap(datos_na.corr(), annot=True, cmap="YlGnBu")
```

```python
# Mostrar las correlaciones de todas las variables con 'median_house_value' ordenadas de mayor a menor
datos_na.corr()["median_house_value"].sort_values(ascending=False)
```
