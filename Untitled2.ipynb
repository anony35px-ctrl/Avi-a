```python
import pandas as pd
datos = pd.read_csv("celsius.csv")
```

```python
import seaborn as sb

sb.scatterplot(x="calsius", y="fahrenheit", data=datos,
               hue="fahrenheit", palette="coolwarm")
```

```python
#Caracteristicas (X), en etiqueta (y)
X = datos['calsius']
y = datos['fahrenheit']
```

```python
X
```

```python
#X.values
#[-40, 10, 0, 8, ...]
#[[-40], [10], [0], [8], ...]

X_procesada = X.values.reshape(-1,1)
y_procesada = y.values.reshape(-1,1)
```

```python
from sklearn.linear_model import LinearRegression

Modelo = LinearRegression()
```

```python
#Entrenamiento
Modelo.fit(X_procesada, y_procesada)
```

```python
calsius = 7900
prediccion = Modelo.predict([[calsius]])
print(f"(calsius) grados calsius som {prediccion} grados fahrenheit")
```

```python
Modelo.score(X_procesada, y_procesada)
```
