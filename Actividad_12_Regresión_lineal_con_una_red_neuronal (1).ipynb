# Regresión lineal utilizando una red neuronal

## Objetivo
Aplicar el algoritmo de regresión lineal mediante una red neuronal para comprender cómo aprende y realiza predicciones.

## Herramientas utilizadas
- Python
- TensorFlow / Keras
- NumPy
- Matplotlib


```python
import pandas as pd

datos = pd.read_csv("celsius.csv")
```

# Parte 1: ¿Cómo funciona una red neuronal?

Una red neuronal aprende patrones ajustando pesos y sesgos para reducir el error entre la predicción y el valor real.

```python
datos.head()
```

```python
datos["celsius"].values
```

## Generación de datos

Se crean los datos de entrenamiento que utilizará la red neuronal.

```python
X = datos ["celsius"].values
y = datos["fahrenheit"].values
```

```python
import tensorflow as tf
modelo = tf.keras.Sequential ([
  tf.keras.layers. Input (shape=(1,)), #la entrada, los celsius: 1 solo dato
  tf.keras.layers.Dense (units=1), #la salida. Tambien es un dato: Los grados fahrenheit
])
modelo.compile(
  optimizer=tf.keras.optimizers.Adam(.1),
  loss="mean_squared_error"
)
```

## Creación de la red neuronal

Se define una red neuronal sencilla de regresión.

```python
entrenamiento = modelo.fit(X, y, epochs=1000)
```

## Entrenamiento del modelo

La red neuronal aprende mediante múltiples épocas ajustando sus parámetros.

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de la pérdida")
plt.plot(entrenamiento.history["loss"])
```

# Parte 2: Red neuronal de regresión

La red neuronal de regresión permite predecir valores numéricos continuos.

```python
import numpy as np
c = 1230
prediccion = modelo.predict(np.array([[c]]))
print(f"Predicción: {c} celsius son {prediccion[0][0]:.2f} fahrenheit")
```

## Resultados

Se comparan las predicciones con los valores reales para evaluar el modelo.

```python
print(modelo.layers[0].get_weights())
```

# Conclusión

La red neuronal logró aprender la relación entre los datos y generar predicciones aproximadas correctamente.
